## Klasifikasi Naive Bayes

Naive Bayes adalah metode klasifikasi berbasis probabilitas yang menggunakan teorema Bayes untuk menentukan kelas dengan peluang terbesar, dengan asumsi setiap atribut saling independen.

## Bayesian Theorema

- Dari training data X, posteriori probabilitas dari hypothesis H atau class, P(H|X), teorema Bayes:


$$
P(C \mid X) = \frac{P(X \mid C)\, P(C)}{P(X)}
$$

- Dapat ditulis dengan:

  **posterior = likelihood × prior / evidence**

- Prediksi X anggota C₂ jika dan hanya jika probabilitas P(C₂|X) paling tinggi di antara semua P(Cₖ|X) dari semua kelas k

## Klasifikasi Naive Bayesian

- Perhatikan D adalah record training dan ditetapkan label-label kelasnya dan masing-masing record dinyatakan n atribut (n field) X = (x₁, x₂, ..., xₙ)

- Misalkan terdapat m kelas C₁, C₂, ..., Cₘ

- Klasifikasi adalah diperoleh maximum posteriori yaitu  
  maksimum P(Cᵢ | X)

- Ini dapat diperoleh dari teorema Bayes:

$$
P(C_i \mid X) = \frac{P(X \mid C_i)\,P(C_i)}{P(X)}
$$

- Karena P(X) adalah konstan untuk semua kelas, hanya perlu memaksimumkan:

$$
P(C_i \mid X) = P(X \mid C_i)\,P(C_i)
$$

## Naive Bayes Classifier

- Diasumsikan: atribut dalam kondisi saling bebas (independent), yaitu tidak ada kebergantungan antara atribut-atribut:

$$
P(X \mid C_i) = \prod_{k=1}^{n} P(x_k \mid C_i) = P(x_1 \mid C_i) \times P(x_2 \mid C_i) \times \cdots \times P(x_n \mid C_i)
$$

- Aₖ adalah categorical, P(xₖ|Cᵢ) adalah jumlah record dalam kelas Cᵢ yang memiliki nilai xₖ sama dengan Aₖ dibagi dengan jumlah record dalam Cᵢ dalam D

- Jika Aₖ bernilai kontinu, P(xₖ|Cᵢ) biasanya dihitung berdasarkan distribusi Gaussian dengan mean (μ) dan standar deviasi σ Dan P(xₖ|Cᵢ) adalah:

$$
g(x, \mu, \sigma) = \frac{1}{\sqrt{2\pi\sigma}} \; e^{-\frac{(x - \mu)^2}{2\sigma^2}}
$$

$$
P(X \mid C_i) = g(x_k, \mu_{C_i}, \sigma_{C_i})
$$
