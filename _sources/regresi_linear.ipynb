## Menggunakan Python Scaler
Selain menggunakan GeoGebra dan Excel, regresi linear juga dapat dihitung menggunakan Python dengan library LinearRegression dari sklearn.linear_model

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# data X dan Y
X = np.array([2,4,5,3,3,4,5]).reshape(-1,1)
Y = np.array([2,3,5,4,3,5,6])

# membuat model regresi
model = LinearRegression()

# training model
model.fit(X, Y)

# menampilkan hasil
print("Slope :", model.coef_[0])
print("Intercept :", model.intercept_)
```
