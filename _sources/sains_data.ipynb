# Sains Data

Mempelajari dan menganalisis data menggunakan alat dan teknik modern untuk menemukan pola yang tidak terlihat dan informasi untuk membuat keputusan bisnis.
Dengan Data Science, data yang awalnya tidak bermakna bisa diubah menjadi dasar untuk pengambilan keputusan yang lebih tepat, cepat, dan cerdas, terutama dalam dunia bisnis.

## Manfaat Sains Data

Berikut beberapa manfaat dari sains data

### 1. Membantu dalam membuat keputusan bisnis

- Membantu perusahaan mencapai tujuan dan memenuhi kepuasan pengguna
- Keputusan berdasarkan data (key business quenstions)
- Credit risk modeling (Proses memperkirakan probalitas)

### 2. Memahami Pelanggan dan Pasar

- Memahami trend permintaan pasar
- Memprediksi kenaikan harga
- Memprediksi penumpangan untuk bisnis transportasi
- Untuk memahami pelanggan 
- Untuk analisa market trend pelanggan bedasarkan

## Seleksi Fitur-Information Gain

Seleksi Fitur: memilih Sebagian fitur (atribut) sedikit mungkin dan memenuhi kebutuhan analisis ( masing infomatif).

Seleksi fitur :
- Menghilangan fitur yang kurang relevan 
- Menghilangkan fitur yang redundan

Merupakan metode mengukur seberapa baik suatu fitur memisahkan data ke dalam kelas-kelas target yang berbeda, dengan memanfaatkan konsep dari teori informasi (information theory), khususnya entropi.
Entropi mengukur tingkat ketidakpastian atau kekacauan dalam suatu dataset.
Entropi tinggi → data sangat bercampur (banyak kelas berbeda).
Entropi rendah → data homogen (hampir semua data milik satu kelas)

### 1. Information Gain

$$ IG(D, A) = H(D) - H(D|A) $$

Keterangan:

- \( H(D) \): Entropi awal dataset sebelum pemisahan  
- \( H(D|A) \): Entropi setelah dataset dibagi berdasarkan fitur \(A\)

### 2. Entropi

$$ H(D) = - \sum_{i=1}^{k} p_i \log_2(p_i) $$

Keterangan:

- \( p_i \): proporsi sampel pada kelas ke-\( i \) dalam dataset \( D \)  
- \( k \): jumlah kelas

### 3. Entropi Bersyarat

$$ H(D|A) = \sum_{v \in nilai(A)} \frac{|D_v|}{|D|} \cdot H(D_v) $$

Keterangan:

- \( D_v \): subset data di mana fitur \( A \) bernilai \( v \)  
- \( \frac{|D_v|}{|D|} \): bobot proporsi dari subset tersebut
