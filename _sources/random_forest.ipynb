# Random Forest

## Analisis Random Forest menggunakan KNIME

![alt text](image18.png)

### 1. CSV Reader

Data yang biasanya masuk di sini berisi:
- Customer ID
- Demografi pelanggan (gender, usia, dll.)
- Data layanan (internet, contract, payment method)
- Label target: Churn (Yes/No)

![alt text](image10.png)

### 2. Table Partitioner

Membagi data menjadi dua bagian:
- Data training (misalnya 70) 
- Data testing (30%)

![alt text](foto10.png)

### 3. Random Forest Lenear

Proses training model machine learning menggunakan algoritma Random Forest.

Cara kerjanya:
- Membuat banyak decision tree dari data training
- Setiap tree belajar pola berbeda dari fitur customer
- Hasil akhir adalah gabungan voting dari semua tree

![alt text](image11.png)
![alt text](image12.png)

### 4. Tree Ensemble Statistic

Menganalisis model yang sudah dilatih.
Biasanya menghasilkan:
- Tingkat pentingnya fitur (feature importance)
- Struktur ensemble tree
- Statistik performa internal model

Pada kasus churn, node ini membantu mengetahui:
fitur apa yang paling mempengaruhi pelanggan berhenti (misalnya contract type, monthly charges, dll.)

![alt text](image13.png)

### 5. Random Forest Prediction

Melakukan prediksi pada data testing.

Prosesnya:
- Model Random Forest yang sudah dilatih digunakan
- Data testing dimasukkan
Output berupa:
- Prediksi churn (Yes/No)
- Probabilitas churn

Hasil ini menunjukkan pelanggan mana yang diprediksi akan berhenti berlangganan.

![alt text](image14.png)

### 6. Scorer

Output yang dihasilkan:
- Accuracy (akurasi model)
- Confusion matrix (TP, TN, FP, FN)
- Precision, recall, dan F1-score (jika diaktifkan)

Pada kasus churn:
- TP: pelanggan churn yang berhasil diprediksi churn
- TN: pelanggan tidak churn yang benar diprediksi tidak churn
- FP/FN: kesalahan prediksi
