# Visualisasi Data Dataset Iris

Pada bagian ini, dilakukan visualisasi untuk memahami distribusi dan hubungan antar variabel numerik pada dataset Iris. Visualisasi membantu melihat pola, perbedaan antar spesies, serta outlier.

```python
iris = load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data['species'] = iris.target

# Map angka ke nama spesies
species_map = {0: 'setosa', 1: 'versicolor', 2: 'virginica'}
data['species'] = data['species'].map(species_map)

data.head()
```

## Boxplot untuk Deteksi Outlier

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=data.drop(columns='species'))
plt.title("Boxplot Variabel Numerik Iris")
plt.show()
```

## Pairplot untuk Hubungan Antar Variabel

```python
sns.pairplot(data, hue='species', diag_kind='hist')
plt.suptitle("Pairplot Variabel Iris per Spesies", y=1.02)
plt.show()
```

### Kesimpulan Visualisasi:

- Variabel **petal length** dan **petal width** sangat membantu membedakan spesies.
- Sepal memiliki overlap antar spesies, lebih sulit dibedakan.
- Tidak terdapat outlier ekstrem pada dataset ini.
- Pola distribusi membantu dalam analisis lanjutan seperti klasifikasi.
