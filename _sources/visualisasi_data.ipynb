# Visualisasi Data

Pada bagian ini, dilakukan visualisasi untuk memahami distribusi dan hubungan antar variabel numerik pada dataset Iris. Visualisasi membantu melihat pola, perbedaan antar spesies, serta outlier.

## Scatter Plot

Scatter plot (diagram pencar) adalah salah satu jenis visualisasi data yang paling mendasar dan kuat dalam dunia statistik dan data science. Visualisasi ini digunakan untuk menunjukkan hubungan atau korelasi antara dua variabel numerik.

## Histogram

Histogram digunakan untuk melihat distribusi dari satu variabel numerik yang membantu untuk mengetahui apakah data tersebut terdistribusi normal atau condong ke satu sisi (skewed).

### Kesimpulan Visualisasi:

- Variabel **petal length** dan **petal width** sangat membantu membedakan spesies.
- Sepal memiliki overlap antar spesies, lebih sulit dibedakan.
- Tidak terdapat outlier ekstrem pada dataset ini.
