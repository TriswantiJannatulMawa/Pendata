# Visualisasi Data

Pada bagian ini, dilakukan visualisasi untuk memahami distribusi dan hubungan antar variabel numerik pada dataset Iris. Visualisasi membantu melihat pola, perbedaan antar spesies, serta outlier.

## Scatter Plot


### Kesimpulan Visualisasi:

- Variabel **petal length** dan **petal width** sangat membantu membedakan spesies.
- Sepal memiliki overlap antar spesies, lebih sulit dibedakan.
- Tidak terdapat outlier ekstrem pada dataset ini.
- Pola distribusi membantu dalam analisis lanjutan seperti klasifikasi.
