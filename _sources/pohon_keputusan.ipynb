# Decision tree

Membangun model klasifikasi menggunakan metode Decision Tree untuk memprediksi apakah seseorang akan bermain tenis (Play Tennis) berdasarkan kondisi cuaca.

Input data

### 2. Table Partitioner

Data kemudian dibagi menjadi dua bagian, yaitu: data training dan data testing

![alt text](foto22.png)

### 3. Decision Tree Learner

Data training digunakan untuk membangun model klasifikasi menggunakan node Decision Tree Learner. Sistem akan menganalisis setiap atribut untuk menentukan atribut terbaik sebagai akar pohon berdasarkan metode tertentu seperti Gain Ratio atau Gini Index. Hasil dari proses ini adalah sebuah model pohon keputusan yang berisi aturan-aturan klasifikasi.

![alt text](foto16.png)
![alt text](foto17.png)

### 4. Decision Tree View

Model yang telah terbentuk kemudian divisualisasikan menggunakan node Decision Tree View. Visualisasi ini menampilkan struktur pohon keputusan secara grafis, sehingga memudahkan pengguna dalam memahami alur pengambilan keputusan yang dihasilkan oleh model.

![alt text](foto-21.png)

### 5. Decision Tree Predictor

Proses ini menghasilkan kolom baru yang berisi hasil prediksi, yaitu apakah data termasuk ke dalam kategori Yes atau No. Tahap ini menunjukkan kemampuan model dalam mengklasifikasikan data yang belum pernah dilihat sebelumnya.

![alt text](foto18.png)

### 6. Scorer

Hasil prediksi dibandingkan dengan data asli untuk mengukur tingkat keakuratan model. Hasil evaluasi biasanya berupa nilai accuracy dan confusion matrix, yang menunjukkan seberapa baik model dalam melakukan klasifikasi. Evaluasi ini penting untuk mengetahui kualitas model yang telah dibangun.

![alt text](foto19.png)
![alt text](foto20.png)

## kesimpulan

Model klasifikasi menggunakan metode Decision Tree untuk memprediksi Play Tennis berdasarkan data yang tersedia. Proses yang dilakukan meliputi pengolahan data, pembagian data, pembuatan model, hingga evaluasi. Hasil yang diperoleh menunjukkan bahwa metode Decision Tree mampu menghasilkan prediksi yang cukup baik serta mudah dipahami melalui bentuk pohon keputusan.
