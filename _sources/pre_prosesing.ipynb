# Pre Prosesing

Preprocessing data adalah tahap awal dalam proses pengolahan data yang bertujuan untuk membersihkan, memperbaiki, dan menyiapkan data mentah agar siap digunakan dalam proses analisis atau data mining.

## Normalization Data

Normalisasi data adalah proses mengubah skala nilai data menjadi rentang tertentu tanpa mengubah informasi utama dari data tersebut.Dengan normalisasi, data yang memiliki nilai sangat besar tidak akan mendominasi atribut lain yang memiliki nilai kecil.

### 1. Min-Max Normalization

Min-Max Normalization digunakan untuk mengubah nilai data ke dalam rentang tertentu, biasanya 0 sampai 1.

$$ v'=\frac{v - min}{max - min} $$
