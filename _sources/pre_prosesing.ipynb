# Pre Prosesing

Preprocessing data adalah tahap awal dalam proses pengolahan data yang bertujuan untuk membersihkan, memperbaiki, dan menyiapkan data mentah agar siap digunakan dalam proses analisis atau data mining.

## Normalization Data

Normalisasi data adalah proses mengubah skala nilai data menjadi rentang tertentu tanpa mengubah informasi utama dari data tersebut.Dengan normalisasi, data yang memiliki nilai sangat besar tidak akan mendominasi atribut lain yang memiliki nilai kecil.

### 1. Min-Max Normalization

Min-Max Normalization digunakan untuk mengubah nilai data ke dalam rentang tertentu, biasanya 0 sampai 1.

$$ v'=\frac{v - min}{max - min} $$

contoh:

| No | Umur | IPK  | Semester |
| -- | ---- | ---- | -------- |
| 1  | 18   | 3.10 | 2        |
| 2  | 19   | 3.25 | 3        |
| 3  | 20   | 3.40 | 4        |
| 4  | 21   | 3.55 | 6        |
| 5  | 22   | 3.70 | 8        |

ipk 1
    min : 3.10
    max : 3.70

$$ v' =\frac{3.10-3.10}{3.70-3.10} = 0 $$

### 2. Z-Score Normalization

Z-score digunakan untuk menormalkan data berdasarkan rata-rata dan standar deviasi sehingga data memiliki mean = 0 dan standar deviasi = 1.

$$ v'=\frac{v - \bar{x}}{\sigma} $$

Contoh data umur:

1. mennetukan mean

$$ \bar{x} = \frac{18+19+20+21+22}{5} = 20 $$

2. menentukan standar deviasi

$$ \sigma = \sqrt{\frac{(18-20)^2 + (19-20)^2 + (20-20)^2 + (21-20)^2 + (22-20)^2}{5}
} = 1.41 $$

3. nilai z-score

$$ v' = \frac{19 - 20}{1.41} = 0.71 $$

### 3. Decimal Scaling Normalization

Decimal scaling menormalkan data dengan membagi nilai dengan pangkat 10 sehingga nilai berada pada rentang kecil.

$$ v'= \frac{v}{10^j} $$

Contoh data semester:
sebelom menghitung nilai kita tentukan terlebih dahulu untuk nilai j, yaitu jumlah digit dari nilai maksimum.
max=8 (1 digit),sehingga nilai j=1

$$ v' = \frac{2}{10^1} = 0.2 $$
