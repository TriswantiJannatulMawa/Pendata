# EKSPLORASI DATA
Eksplorasi data adalah proses awal dalam analisis data yang bertujuan untuk memahami karakteristik, pola, dan struktur dari suatu kumpulan data sebelum dilakukan analisis lebih lanjut atau pembuatan model. Dengan melakukan eksplorasi data, analis dapat memahami kondisi data secara menyeluruh sehingga dapat menentukan metode analisis yang tepat dan menghasilkan kesimpulan yang lebih akurat.
Dataset ini memiliki 150 data dengan 4 variabel numerik:
- Sepal Length
- Sepal Width
- Petal Length
- Petal Width
Serta 1 variabel target yaitu spesies bunga.

jumlah Masing-masing data:
- setosa : 50 data
