## EKSPLORASI DATA
Eksplorasi data adalah proses awal dalam analisis data yang bertujuan untuk memahami karakteristik, pola, dan struktur dari suatu kumpulan data sebelum dilakukan analisis lebih lanjut atau pembuatan model. Pada tahap ini, peneliti atau analis mencoba mengenali isi data dengan melihat jumlah dan jenis variabel, tipe data, distribusi nilai, serta hubungan antarvariabel. Eksplorasi data juga digunakan untuk mendeteksi adanya data yang hilang, data duplikat, atau nilai yang tidak wajar (outlier) yang dapat memengaruhi hasil analisis. Proses ini biasanya melibatkan perhitungan statistik deskriptif seperti rata-rata, median, dan standar deviasi, serta penggunaan visualisasi seperti grafik atau diagram untuk membantu melihat pola secara lebih jelas. Dengan melakukan eksplorasi data, analis dapat memahami kondisi data secara menyeluruh sehingga dapat menentukan metode analisis yang tepat dan menghasilkan kesimpulan yang lebih akurat.

Dataset ini memiliki 150 data dengan 4 variabel numerik:

- Sepal Length
- Sepal Width
- Petal Length
- Petal Width
Serta 1 variabel target yaitu spesies bunga.

## Menampilkan 5 data pertama

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# Load dataset
iris = load_iris()
data = pd.DataFrame(iris.data, columns=iris.feature_names)
data['species'] = iris.target
```

## Distribusi Variabel

Kita akan melihat distribusi masing-masing variabel numerik menggunakan histogram.

```python
data.hist(figsize=(10,8), bins=15, edgecolor='black')
plt.suptitle('Histogram Variabel Iris', fontsize=16)
plt.show()
```
