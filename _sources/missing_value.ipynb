# Missing Value

Missing value adalah kondisi ketika suatu data tidak memiliki nilai atau kosong pada suatu atribut dalam dataset. Missing value biasanya muncul karena beberapa faktor seperti kesalahan pencatatan, data yang tidak lengkap, atau kesalahan saat proses pengumpulan data.

## Menangani Missing value

Ada beberapa metode untuk menangani missing value

1. Menghapus data (deletion)
Yaitu dengan menghapus data yang memiliki nilai kosong dan cocok jika jumlah missing value kecil

2. Mengisi dengan Mean,
Metode ini digunakan jika data berupa numerik

3. Mengisi dengan Median,
Metode ini digunakan jika data memiliki outlier

4. Mengisi dengan Modus,
Digunkan untuk data yang katagorikal

5. Interpolasi,
Mengisi nilai yang hilang berdasarkan pola data

## Tugas Penanganan Missing Value dengan WKNN

| No  | IPK | PO      | JML  |
| --- | --- | ------- | -----|
| 1   | 2   | 2000000 | 2    |
| 2   | 3   | 3000000 | 3    |

### 1. Hitung jarak euclidean data ke 7
| No  | minmax IPK | minmax PO | minmax JML  |
| --- | --- | ------- | -----|
| 1   | 0   | 0   | 0    |
| 2   | 0.5 | 0.5 | 1    |
| 3   | 1   | 0   | 0    |
| 4   | 0   | 0   | 1    |
| 5   | 0.5 | 0.5 | 0    |
| 6   | 1   | 1   | 1    |
| 7   | 0   | 0.5 | ?    |

$$ d(x,y) = \sqrt{(IPK_x-IPK_y)^2 + (PO_x-PO_y)^2} $$

$$ d(1,7) = \sqrt{(0-0)^2 + (0-0.5)^2} = 0.5 $$

$$ d(2,7) = \sqrt{(0.5-0)^2 + (0.5-0.5)^2} = 0.5 $$

$$ d(3,7) = \sqrt{(1-0)^2 + (0-0.5)^2} = 1.118 $$

$$ d(4,7) = \sqrt{(0-0)^2 + (0-0.5)^2} = 0.5 $$

$$ d(5,7) = \sqrt{(0.5-0)^2 + (0.5-0.5)^2} = 0.5 $$

$$ d(6,7) = \sqrt{(1-0)^2 + (1-0.5)^2} = 1.118 $$


### 2. Menentukan 3 tetangga terdekat
dari data diatas terdapat 4 data yang nilainya sama maka kita pilih 3 teratas yaitu data ke 1,2 dan 4.
- d(1) = 0.5 (JML=0)
- d(2) = 0.5 (JML=1)
- d(4) = 0.5 (JML=1)

### 3. Hitung bobot

$ w = \frac{1}{d^2} $ 

$ w = \frac{1}{0.5^2} = 4 $

### 4. Prediksi nilai JML

$$ JML = \frac{(w \cdot JML_1) + (w \cdot JML_2) + (w \cdot JML_4) }{w + W + w} $$

$$ JML = \frac{(4 \cdot 0) + (4 \cdot 1) + (4 \cdot 1) }{4 + 4 + 4} = 0.67 $$

Prediksi akhir JML data ke 7 yaitu 0.67

## Implementasi Code 

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

# Data yang sudah ternormalisasi
data_norm = {
    'IPK': [0, 0.5, 1, 0, 0.5, 1, 0],
    'PO': [0, 0.5, 0, 0, 0.5, 1, 0.5],
    'JML': [0, 1, 0, 1, 0, 1, np.nan] 
}

df = pd.DataFrame(data_norm)

# Inisialisasi WKNN dengan k=3
imputer = KNNImputer(n_neighbors=3, weights='distance')

# Lakukan Imputasi
df_imputed = pd.DataFrame(imputer.fit_transform(df), columns=df.columns)

print("Hasil Imputasi JML baris ke-7:")
print(df_imputed.iloc[[6]])
```
